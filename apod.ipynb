# Astronomy picture of the day

This notebook loads a NASA API key from a `.env` file with `python-dotenv`, then uses it to fetch and display NASA's [astronomy picture of the day](https://apod.nasa.gov/apod/astropix.html) (APOD).

This notebook was generated using AI under general human direction. At the time of generation, the contents have not been comprehensively reviewed by a human analyst.

```python
# load-packages
import os

import requests
from dotenv import load_dotenv
from IPython.display import Image, Markdown, display
```

## Load the API key

The key lives in `.env` (which is gitignored, so it never gets committed). `load_dotenv()` reads that file and populates `os.environ`, so the notebook code never has to contain the secret itself.

```python
# load-api-key
load_dotenv()

api_key = os.environ["NASA_API_KEY"]  # fails loudly if .env is missing the key
```

## Fetch the picture of the day

The [APOD API](https://api.nasa.gov/) takes the key as a query parameter, and `raise_for_status()` turns a failed request (bad key, rate limit, server error) into a visible error.

Note: NASA's APOD service is mid-migration right now and is flaky — it sometimes returns a 500 or a generic logo placeholder. If that happens, just rerun the cell.

```python
# fetch-apod
response = requests.get(
    "https://api.nasa.gov/planetary/apod",
    params={"api_key": api_key},
)
response.raise_for_status()
apod = response.json()

# A 200 response can still be a placeholder rather than today's picture.
if apod["title"] == "NASA Science":
    raise RuntimeError("The API served a placeholder — rerun this cell.")

apod
```

## Display the picture

The JSON response has the `title`, `date`, `explanation`, and the image URL in `hdurl`. IPython's `display` helpers render each one in the notebook.

```python
# display-apod
display(Markdown(f"## {apod['title']}"))
display(Markdown(f"*{apod['date']}*"))
display(Image(url=apod.get("hdurl") or apod["url"]))
display(Markdown(apod["explanation"]))
```
